```python
import os
import pandas as pd
```

```python
# Step 1: Auto-detect project root
# ------------------------------------
current_dir = os.getcwd()
project_root = os.path.abspath(os.path.join(current_dir, ".."))

# Fallback if already at project root
if not os.path.exists(os.path.join(project_root, "data")):
    project_root = current_dir

print("✅ Project root:", project_root)
```

```python
# Step 2: Load processed cross-platform dataset
# ------------------------------------
processed_path = os.path.join(
    project_root,
    "data",
    "processed",
    "cross_platform_dataset_processed.csv"
)

df_processed = pd.read_csv(processed_path)

print("✅ Processed dataset loaded")
print("Columns:", list(df_processed.columns))
```

```python
# Step 3: Load existing cross-platform test split
# ------------------------------------
X_test_cross_path = os.path.join(
    project_root,
    "data",
    "processed",
    "splits",
    "X_test_cross.csv"
)

X_test_cross = pd.read_csv(X_test_cross_path)

print("✅ X_test_cross loaded")
print("Rows in X_test_cross:", len(X_test_cross))
```

```python
# Step 4: Align rows (CRITICAL STEP)
# ------------------------------------
df_test_cross = df_processed.loc[X_test_cross.index].copy()

print("✅ Rows aligned")
print("Rows in df_test_cross:", len(df_test_cross))
```

```python
# Step 5: Final sanity checks
# ------------------------------------
assert len(df_test_cross) == len(X_test_cross), "❌ Row mismatch error!"

print("Platforms present:", df_test_cross['Platform'].unique())
```

```python
# Step 6: Save final CSV with platform info
# ------------------------------------
save_path = os.path.join(
    project_root,
    "data",
    "processed",
    "splits",
    "X_test_cross_full.csv"
)

df_test_cross[['clean_text', 'label_encoded', 'Platform']].to_csv(
    save_path,
    index=False
)

print("✅ X_test_cross_full.csv created successfully")
print("📁 Saved at:", save_path)
```
