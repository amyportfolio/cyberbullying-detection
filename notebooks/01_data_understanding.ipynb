DATASET UNDERSTANDING

STEP 1 - CROSS-PLATFORM DATASET (KAGGLE)

```python
import os
import pandas as pd
```

```python
# Detect project root dynamically
current_dir = os.getcwd()
project_root = os.path.abspath(os.path.join(current_dir, ".."))

if not os.path.exists(os.path.join(project_root, "data")):
    project_root = current_dir

print("Project root detected:", project_root)
```

```python
# Load dataset
raw_data_path = os.path.join(
    project_root, "data", "raw", "cross_platform_dataset.csv"
)

print("File exists:", os.path.exists(raw_data_path))

df = pd.read_csv(raw_data_path)
```

```python
def dataset_summary(df, name):
    print(f"\n📌 Dataset: {name}")
    print("Shape:", df.shape)

    print("Number of rows:", df.shape[0])
    print("Number of columns:", df.shape[1])

    print("\nColumns:")
    print(df.dtypes)
    
    print("\nMissing values:")
    print(df.isnull().sum())
```

```python
# Call function
dataset_summary(df, "Cross-Platform Cyberbullying Dataset (Raw)")
```

```python
print(df["Platform"].value_counts())
print(df["Target"].value_counts())
```

```python
# Show 5 random samples
df.sample(5)[["Platform", "Text_Content", "Target"]]
```

```python
# Rename 'Neutral' to 'not_cyberbullying' in Dataset 1
df["Target"] = df["Target"].apply(lambda x: "Cyberbullying" if x == "Cyberbullying" else "not_cyberbullying")

# Quick check
df["Target"].value_counts()
```

```python
# Add text length column
df["text_length"] = df["Text_Content"].apply(len)

# Compare average lengths
df.groupby("Target")["text_length"].describe()
```

STEP 2 - TWITTER CYBERBULLYING DATASET (KAGGLE)

```python
import os
import pandas as pd

# Detect project root
current_dir = os.getcwd()
project_root = os.path.abspath(os.path.join(current_dir, ".."))

if not os.path.exists(os.path.join(project_root, "data")):
    project_root = current_dir

print("Project root:", project_root)

# Kaggle dataset path
kaggle_path = os.path.join(
    project_root, "data", "raw", "kaggle_dataset.csv"
)

print("File exists:", os.path.exists(kaggle_path))

# Load dataset
df_kaggle = pd.read_csv(kaggle_path)

# Basic info
print("\nShape:", df_kaggle.shape)
df_kaggle.head()
```

```python
def dataset_summary(df, name):
    print(f"\n📌 Dataset: {name}")
    print("Shape:", df.shape)
    print("\nColumns & Data Types:")
    print(df.dtypes)
    print("\nMissing values:")
    print(df.isnull().sum())

dataset_summary(df_kaggle, "Kaggle Twitter Cyberbullying Dataset")
```

```python
#Check Unique Labels
df_kaggle["cyberbullying_type"].value_counts()
```

```python
# Binary mapping with 'not_cyberbullying' as the neutral label
df_kaggle["Target"] = df_kaggle["cyberbullying_type"].apply(
    lambda x: "Cyberbullying" if x != "not_cyberbullying" else "not_cyberbullying"
)

# Quick check
print(df_kaggle["Target"].value_counts())
```

```python
# Show 5 random samples
df_kaggle.sample(5)[["tweet_text", "Target"]]
```

```python
# Add text length column
df_kaggle["text_length"] = df_kaggle["tweet_text"].apply(len)

# Compare lengths by label
df_kaggle.groupby("Target")["text_length"].describe()
```

STEP 3 - Twitter API fetched

```python
import os
import pandas as pd

# Detect project root
current_dir = os.getcwd()
project_root = os.path.abspath(os.path.join(current_dir, ".."))

if not os.path.exists(os.path.join(project_root, "data")):
    project_root = current_dir

# Dataset 3 path
twitter_path = os.path.join(project_root, "data", "raw", "twitter.csv")
print("File exists:", os.path.exists(twitter_path))

# Load dataset
df_twitter = pd.read_csv(twitter_path)

# Basic info
print("\nShape:", df_twitter.shape)
df_twitter.head()
```

```python
def dataset_summary(df, name):
    print(f"\n📌 Dataset: {name}")
    print("Shape:", df.shape)
    print("\nColumns & Data Types:")
    print(df.dtypes)
    print("\nMissing values:")
    print(df.isnull().sum())

dataset_summary(df_twitter, "Twitter API Fetched Dataset")
```

```python
df_twitter["label"].value_counts()
```

```python
# Map Labels to Same Format as Other Datasets
df_twitter["Target"] = df_twitter["label"].apply(lambda x: "Cyberbullying" if x == 1 else "not_cyberbullying")

# Quick check
df_twitter["Target"].value_counts()
```

```python
# Inspect Text Samples
df_twitter.sample(5)[["platform", "text", "Target"]]
```

```python
# Text Length Analysis

# Add text length column
df_twitter["text_length"] = df_twitter["text"].apply(len)

# Compare lengths by label
df_twitter.groupby("Target")["text_length"].describe()
```
