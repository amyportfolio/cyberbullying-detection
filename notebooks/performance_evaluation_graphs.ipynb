## Accuracy Comparison on Twitter Dataset

```python
# Replace these values with actual results

models = [
    "Logistic\nRegression",
    "LSTM",
    "CNN-BiLSTM",
    "BERT",
    "DistilBERT"
]

twitter_accuracy = [
    84.96,
    86.00,
    82.33,
    84.00,
    84.00
]

plt.figure(figsize=(8,5))

bars = plt.bar(models, twitter_accuracy)

plt.title("Accuracy Comparison on Twitter Dataset")
plt.xlabel("Models")
plt.ylabel("Accuracy (%)")
plt.ylim(0,100)

for bar in bars:
    height = bar.get_height()
    plt.text(
        bar.get_x()+bar.get_width()/2,
        height+0.5,
        f"{height:.2f}",
        ha='center'
    )

plt.tight_layout()

plt.savefig("twitter_accuracy.png", dpi=300)

plt.show()
```

## Platform-wise Accuracy Comparison

```python
import numpy as np
import matplotlib.pyplot as plt

# Models
models = [
    "Logistic\nRegression",
    "LSTM",
    "CNN-BiLSTM",
    "BERT",
    "DistilBERT"
]

# Replace these with your actual accuracy values
twitter = [
    84.96,
    86.00,
    82.33,
    84.00,
    84.00
]

facebook = [
    69,
    75,
    77,
    80,
    81
]

instagram = [
    70,
    76,
    78,
    81,
    82
]

x = np.arange(len(models))
width = 0.25

plt.figure(figsize=(10,6))

bars1 = plt.bar(x - width, twitter, width, label="Twitter")
bars2 = plt.bar(x, facebook, width, label="Facebook")
bars3 = plt.bar(x + width, instagram, width, label="Instagram")

# Display values on bars
for bars in [bars1, bars2, bars3]:
    for bar in bars:
        height = bar.get_height()
        plt.text(
            bar.get_x() + bar.get_width()/2,
            height + 0.5,
            f"{height:.1f}",
            ha='center',
            fontsize=8
        )

plt.xticks(x, models)

plt.xlabel("Models")
plt.ylabel("Accuracy (%)")
plt.title("Platform-wise Accuracy Comparison")

plt.ylim(0, 100)

plt.legend()

plt.grid(axis='y', linestyle='--', alpha=0.5)

plt.tight_layout()

plt.savefig("platform_accuracy.png", dpi=300)

plt.show()
```
