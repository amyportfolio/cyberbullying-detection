```python
import os
import pandas as pd
```

Detect project root dynamically

```python
# Assume project root is the folder containing 'data' folder
current_dir = os.getcwd()
project_root = os.path.abspath(os.path.join(current_dir, ".."))  # one level up from notebooks

# If running from project root, this will still work
if not os.path.exists(os.path.join(project_root, "data")):
    project_root = current_dir  # fallback if already at root

print("Project root detected:", project_root)
```

Load dataset

```python
raw_data_path = os.path.join(project_root, "data", "resources", "cyberbullying_dataset.csv")
df = pd.read_csv(raw_data_path)
```

```python
# Basic checks
print("Total rows:", len(df))
print("\nPlatforms present:")
print(df["Platform"].value_counts())
```

```python
df.head()
```

```python
# Remove Reddit
allowed_platforms = ["Reddit", "Facebook", "Instagram"]
df_filtered = df[df["Platform"].isin(allowed_platforms)]

print("\nAfter removing Twitter:")
print(df_filtered["Platform"].value_counts())
```

```python
# Keep only required columns
df_filtered = df_filtered[["Platform", "Text_Content", "Target", "Timestamp"]]

# Check the first few rows
print(df_filtered.head())
```

Ensure processed folder exists

```python
folder = os.path.join(project_root, "data", "raw")
os.makedirs(folder, exist_ok=True)
```

Save filtered dataset

```python
file_path = os.path.join(folder, "cross_platform_dataset.csv")
df_filtered.to_csv(file_path, index=False)

print("\n✅ Clean dataset saved successfully at:", file_path)
```
